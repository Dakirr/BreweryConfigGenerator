```python
def pale_hex_color(hex_color, strength=0.5):
    hex_color = hex_color.lstrip('#')
    
    r = int(hex_color[0:2], 16)
    g = int(hex_color[2:4], 16)
    b = int(hex_color[4:6], 16)
    
    r = int(r + (255 - r) * strength)
    g = int(g + (255 - g) * strength)
    b = int(b + (255 - b) * strength)
    
    return f"#{r:02x}{g:02x}{b:02x}"

```

```python
def get_ingredients_text(ingredients):
    ret = "ingredients:\n"
    for i in ingredients:
        ret += f"      - {i[0]}/{min(max(round(i[1] + random.choice([-2, -1, -1, 0, 1, 1])), round(i[1]*(1 - VARIANCE)), 1), round(i[1]*(1 + VARIANCE)))}\n"
    return ret.strip("\n")

def get_effects_text(positive, negative):
    ret = "effects:\n"
    ret += f"      - HUNGER/200/1\n"
    for i in positive:
        ret += f"      - {i[0]}/0-{i[2]}/0-{i[1]}\n"
    for i in negative:
        ret += f"      - {i[0]}/{i[2]}-0/{i[1]}-0\n"
    return ret.strip("\n")

CustomIdCounter = 556

class Recipe:
    def __init__(self, name, color = None, alcohol = None, color_preffix = "", positive_preffix = None, negative_preffix = None, postfix="dk", glint = False):
        self.name = name
        self.color_preffix = color_preffix
        if (not positive_preffix is None): self.positive_preffix = positive_preffix
        else: self.positive_preffix = random.choice(positive_preffixes)
        if (not negative_preffix is None): self.negative_preffix = negative_preffix
        else: self.negative_preffix = random.choice(negative_preffixes)
        self.color = color
        if (not alcohol is None): self.alcohol = alcohol
        else: self.alcohol = 20 + 5*random.randint(-1,1)
        global CustomIdCounter
        self.CID = CustomIdCounter
        CustomIdCounter += 1
        self.postfix = postfix
        self.add_aging_info()
        self.add_distill_info()
        self.add_lore()
        self.add_drinkmessages()
        self.glint = glint
        

    def add_cooking_info(self, ingredients, difficulty = None, cooking_time = None, generate_effects = True):
        if (not difficulty is None): self.difficulty = difficulty
        else: self.difficulty = 7
        self.ingredients = get_ingredients_text(ingredients)
        self._ilist = ingredients
        if (not cooking_time is None): self.cooking_time = cooking_time
        else: self.cooking_time = 5 + random.choice([-3,-3,-2,-1,-1,0,0,1,2,3])
        if generate_effects:
            self.add_effects()

    def add_aging_info(self, age_time = 0, age_wood = 0):
        if (age_time):
            self.age_time = max(age_time + random.choice([-2,-1,0,0,0,1,1,2]), 1)
            self.age_wood = age_wood
        else:
            self.age_time = 0
            self.age_wood = age_wood = 0

    def add_distill_info(self, distill_runs = 0, distill_time=40):
        if (distill_runs):
            self.distill_runs = max(distill_runs + random.choice([-2,-1,0,0,0,1,1,2]), 1)
            self.distill_time = distill_time
        else:
            self.distill_runs = 0
            self.distill_time = 40

    def add_lore(self, positive_lore = "", neutral_lore = "", negative_lore = ""):
        self.positive_lore = positive_lore
        self.neutral_lore = neutral_lore
        self.negative_lore = negative_lore

    def add_drinkmessages(self, drinkmessage = "", drinktitle = ""):
        self.drinkmessage = drinkmessage
        self.drinktitle = drinktitle

    def add_effects(self, effects_text = ""):
        if (effects_text != ""): self.effects = effects_text
        else:
            self.effects_positive = []
            self.effects_negative = []
            main_positive_effect_type = ingredients_map[self._ilist[0][0]][3]
            self.effects_positive.append(random.choice(positive_effects_reference[main_positive_effect_type][1]))
            self.effects_positive.append(random.choice(positive_effects_reference[main_positive_effect_type][0]))
            secondary_positive_effect_type = ingredients_map[self._ilist[1][0]][3]
            t = random.choice(positive_effects_reference[secondary_positive_effect_type][0])
            if (self.effects_positive[-1] != t):
                self.effects_positive.append(t)
            

            main_negative_effect_type = ingredients_map[self._ilist[0][0]][4]
            self.effects_negative.append(random.choice(negative_effects_reference[main_negative_effect_type][1]))
            secondary_negative_effect_type = ingredients_map[self._ilist[1][0]][4]
            self.effects_negative.append(random.choice(negative_effects_reference[secondary_negative_effect_type][0]))
            if len(self._ilist) >= 3:
                secondary_negative_effect_type = ingredients_map[self._ilist[2][0]][4]
                t = random.choice(negative_effects_reference[secondary_negative_effect_type][0])
                if (self.effects_negative[-1] != t):
                    self.effects_negative.append(t)
                self.effects = get_effects_text(self.effects_positive, self.effects_negative)
# self, name, positive_preffix, negative_preffix, color, difficulty, ingredients, alcohol, brew_time, age_wood, age_time, distill_runs, distill_time=40, drinkmessage = "", drinktitle = ""

    def get_recipe_text_brewery(self):
        recipe = f"""{self.name.lower().replace(" ", "_")}_{self.postfix.lower()}:
    name: {self.negative_preffix} {self.name.lower()}/{self.name}/{self.color_preffix}{self.positive_preffix} {self.name.lower()}
    {self.ingredients}
    cookingtime: {self.cooking_time}
    distillruns: {self.distill_runs}\n"""
        if self.distill_runs != 0:
            recipe += f"    distilltime: {self.distill_time}\n"
        if self.age_time != 0: recipe += f"    wood: {self.age_wood}\n"
        recipe += f'    age: {self.age_time}'
        recipe += f"""
    color: '{self.color[1:]}'
    difficulty: {self.difficulty}
    alcohol: {self.alcohol}
    """
        if self.positive_lore != "" or self.neutral_lore != "" or self.negative_lore != "":
            recipe += "lore:\n"
            if (self.positive_lore != ""):
                recipe += f"      - +++{self.positive_lore}\n"
            if (self.neutral_lore != ""):
                recipe += f"      - ++{self.neutral_lore}\n"
            if (self.negative_lore != ""):
                recipe += f"      - +{self.negative_lore}\n"
        # servercommands:
        # - +++ weather clear
        # - + weather rain
        # playercommands:
        # - homes
        if (self.drinkmessage != ""):
            recipe += f"    drinkmessage: {self.drinkmessage}\n"
        if (self.drinktitle != ""):
            recipe += f"    drinktitle: {self.drinktitle}\n"
        recipe += f"    customModelData: {self.CID}/{self.CID}/{self.CID}\n"
        if (self.effects != ""):
            recipe += "    " + self.effects
        return recipe
    
    def get_recipe_text_breweryX(self):
        recipe = f"""{self.name.lower().replace(" ", "_")}_{self.postfix.lower()}:
    name: {self.negative_preffix} {self.name.lower()}/{self.name}/&{pale_hex_color(self.color, 0.2)}{self.positive_preffix} {self.name.lower()}
    {self.ingredients}
    cookingtime: {self.cooking_time}
    distillruns: {self.distill_runs}\n"""
        if self.distill_runs != 0:
            recipe += f"    distilltime: {self.distill_time}\n"
        if self.age_time != 0: recipe += f"    wood: {self.age_wood}\n"
        recipe += f'    age: {self.age_time}'
        recipe += f"""
    color: '{self.color[1:]}'
    difficulty: {self.difficulty}
    alcohol: {self.alcohol}
    """
        if self.positive_lore != "" or self.neutral_lore != "" or self.negative_lore != "":
            recipe += "lore:\n"
            if (self.positive_lore != ""):
                recipe += f"      - +++&{pale_hex_color(self.color, 0.4)}{self.positive_lore}\n"
            if (self.neutral_lore != ""):
                recipe += f"      - ++&{pale_hex_color(self.color, 0.4)}{self.neutral_lore}\n"
            if (self.negative_lore != ""):
                recipe += f"      - +&{pale_hex_color(self.color, 0.4)}{self.negative_lore}\n"
        # servercommands:
        # - +++ weather clear
        # - + weather rain
        # playercommands:
        # - homes
        if (self.glint):
            recipe += "    glint: true\n"
        if (self.drinkmessage != ""):
            recipe += f"    drinkmessage: {self.drinkmessage}\n"
        if (self.drinktitle != ""):
            recipe += f"    drinktitle: {self.drinktitle}\n"
        recipe += f"    customModelData: {self.CID}/{self.CID}/{self.CID}\n"
        if (self.effects != ""):
            recipe += "    " + self.effects
        return recipe
```

```python
def get_brew_text(ingridient_name, ingridient_color):
    return f"""{ingridient_name.lower().replace(" ", "_")}_brew:
    name: Brewed {ingridient_name.replace("_", " ").lower()}
    ingredients: {ingridient_name}
    color: '{ingridient_color}'
    cookParticles:
    - '{ingridient_color[1:]}/1'
    - '{pale_hex_color(ingridient_color, 0.1)[1:]}/2'
    - '{pale_hex_color(ingridient_color, 0.3)[1:]}/4'
    - '{pale_hex_color(ingridient_color, 0.5)[1:]}/6'
    - '{pale_hex_color(ingridient_color, 0.7)[1:]}/8'
    """
```

```python
import os
import shutil
```

```python
class ConfigCreator:
    def __init__(self):
        self.recipes = set()
        self.brews = set()
        self.config = ""
        
    def add_recipe(self, recipe: Recipe):
        self.recipes.add(recipe)
        for i in recipe._ilist:
            self.brews.add((i[0], ingredients_map[i[0]][2]))

    def finalize_brewery(self):        
        with open('res/base_brewery.yml', 'r') as file:
            self.config += file.read()
        shutil.rmtree("output")
        os.mkdir("output")
        for i in self.brews:
            for j in get_brew_text(i[0], i[1]).split("\n"):
                self.config += "  " + j
                self.config += "\n"
        
        self.config += "recipes:\n"
        for i in self.recipes:
            for j in i.get_recipe_text_brewery().split("\n"):
                self.config += "  " + j
                self.config += "\n"
            self.config += "\n"
        
        with open('res/ending_brewery.yml', 'r') as file:
            self.config += file.read()

        with open('output/config.yml', 'w') as file:
            file.write(self.config)

    def finalize_breweryX(self):
        shutil.rmtree("output")
        os.mkdir("output")

        self.config = ""
        with open('res/config_breweryX.yml', 'r') as file:
            self.config += file.read()
        with open('output/config.yml', 'w') as file:
            file.write(self.config)

        self.config = ""
        with open('res/custom-items_breweryX.yml', 'r') as file:
            self.config += file.read()
        with open('output/custom-items.yml', 'w') as file:
            file.write(self.config)

        self.config = ""
        self.config += "cauldron:\n"
        for i in self.brews:
            for j in get_brew_text(i[0], i[1]).split("\n"):
                self.config += "  " + j
                self.config += "\n"
        with open("output/cauldron.yml", 'w') as file:
            file.write(self.config)

        self.config = ""
        self.config += "recipes:\n"
        for i in self.recipes:
            for j in i.get_recipe_text_breweryX().split("\n"):
                self.config += "  " + j
                self.config += "\n"
            self.config += "\n"
        with open('output/recipes.yml', 'w') as file:
            file.write(self.config)
```
